# Rainfall and yield, 1990 to 2024

Pulls the county rainfall series against reported yield and looks for a lag.

```python
import pandas as pd
import matplotlib.pyplot as plt

rain = pd.read_csv("data/rainfall.csv", parse_dates=["date"])
yields = pd.read_csv("data/yields.csv")
rain.head()
```

```python
merged = rain.groupby(rain.date.dt.year).total.sum().to_frame().join(
    yields.set_index("year")
)
merged.plot(subplots=True, figsize=(9, 5))
plt.tight_layout()
```
